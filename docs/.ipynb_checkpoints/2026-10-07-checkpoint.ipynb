```python
import matplotlib.pyplot as plt
import numpy as np

A = np.array([[1, 2], [2, 5]])
B = np.array([4, 9])

solution = np.linalg.solve(A, B)
print("연립방정식의 해: ", solution)


x = np.linspace(-1, 6, 400)  # -1부터 6까지 400개의 점을 생성
y1 = (4-x) / 2
y2 = (9 - 2*x) / 5

plt.figure(figsize=(8, 6))
plt.plot(x, y1, label="y1", linewidth=2, color="blue")
plt.plot(x, y2, label="y2", linewidth=2, color="red")
plt.scatter(solution[0], solution[1], color="black", s=100, label="solution")  # s는 사이즈
plt.legend()
plt.grid(True)
plt.show()
```

```python
A = np.arange(6).reshape(3, 2)
B = np.arange(6).reshape(3, 2) - 3

print(A)
print("전치행렬: \n", A.T)  # 전치행렬
print(A.transpose())
print("-" * 40)
print(B)

print("-" * 40)
print("transpose")
print(B.transpose())
print("-" * 40)

print(A * B)  # element-wise multiplication: dot product와 다르다
print(A.dot(B.transpose()))  # matrix multiplication (행렬 곱)

print("-" * 40)

A = np.arange(6).reshape(2, 3)
B = np.arange(-5, 1).reshape(3, 2)
C = np.dot(A, B)
print(C)
```

```python
A = np.arange(6).reshape(2, 3)
B = np.arange(-5, 1).reshape(2, 3)
print(A)
print(B)

print("-" * 40)
print("A + B:\n", A + B)

print("-" * 40)
print("A - B:\n", A - B)

print("-" * 40)
print("A * B:\n", A * B)  # 원소별 곱(dot product가 아닌 element-wise multiplication)
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42576?language=python3

def solution(participant: list[str], completion: list[str]) -> str:
    """
    https://bunnydev-jordy.tistory.com/entry/파이썬Python-프로그래머스-42576-완주하지-못한-선수-해시-딕셔너리-카운터
    https://hangjastar.tistory.com/254
    """

    answer = ""
    
    count: dict[str, int] = {}
    
    for name in participant:
        count[name] = count.get(name, 0) + 1

    for name in completion:
        count[name] -= 1

    for name, cnt in count.items():
        if cnt:
            answer = name
    
    return answer


print(solution(["leo", "kiki", "eden"], ["eden", "kiki"]))  # "leo"
print(
    solution(
        ["marina", "josipa", "nikola", "vinko", "filipa"],
        ["josipa", "filipa", "marina", "nikola"],
    )
)  # "vinko"
print(solution(["mislav", "stanko", "mislav", "ana"], ["stanko", "ana", "mislav"]))  # 	"mislav"
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42576?language=python3

def solution(participant: list[str], completion: list[str]) -> str:
    """
    https://unie2.tistory.com/1070
    """
    
    answer = ""
    participant.sort()
    completion.sort()

    for p, c in zip(participant, completion):
        print("참가자, 완주자", p, c)
        if p != c:
            return p

    answer = participant[-1]
    
    return answer


print(solution(["leo", "kiki", "eden"], ["eden", "kiki"]))  # "leo"
print(
    solution(
        ["marina", "josipa", "nikola", "vinko", "filipa"],
        ["josipa", "filipa", "marina", "nikola"],
    )
)  # "vinko"
print(solution(["mislav", "stanko", "mislav", "ana"], ["stanko", "ana", "mislav"]))  # 	"mislav"
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181917

def solution(x1: bool, x2: bool, x3: bool, x4: bool) -> bool:
    answer = True

    return (x1 or x2) and (x3 or x4)
    
    return answer


print(solution(False, True, True, True))  # True
print(solution(True, False, False, False))  # False
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181916?language=python3

def solution(a: int, b: int, c: int, d: int) -> int:
    answer = 0

    if a == b == c == d:
        return str(a) * 4
    
    return answer


print(solution(2, 2, 2, 2))  # 2222
print(solution(4, 1, 4, 4))  # 1681
print(solution(6, 3, 3, 6))  # 27
print(solution(2, 5, 2, 6))  # 30
print(solution(6, 4, 2, 5))  # 2
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181915?language=python3

def solution(my_string: str, index_list: list[int]) -> str:
    answer = ''

    for idx in index_list:
        answer += my_string[idx]
    
    return answer


print(solution("cvsgiorszzzmrpaqpe", [16, 6, 5, 3, 12, 14, 11, 11, 17, 12, 7]))  # "programmers"
print(solution("zpiaz", [1, 2, 0, 0, 3]))  # "programmers"
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181913?language=python3

def solution(my_string: str, queries: list[list[int]]) -> str:
    answer = ''

    partial = ""

    for query in queries:
        s, e = query
        
        for my_str in my_string[s:e+1]:
            partial += my_str

        my_str = my_str[::-1]
        print(my_str)
    
    return answer


print(solution("rermgorpsam", [[2, 3], [0, 7], [5, 9], [6, 10]]))  # "programmers"
```
