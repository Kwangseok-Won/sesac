```python
def treeprinter(n: int):
    for i in range(n):
        if i < n - 5:
            print(" " * (n-i) + "*" * (2*i+1))
        else:
            print(" " * (n-1) + "*" * 3)

treeprinter(25)
```

데코레이터 - closure의 특수한 형태

```python
import time


def decorator(func):
    def wrapper(*args, **kwargs):
        print("Start")

        result = func(*args, **kwargs)

        print("End")

        return result
    return wrapper

@decorator
def add(x, y):
    return x + y


print(add(3, 5))

print("-" * 40)


def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()
        print(f"실행 시간: {end - start:.4f}초")

        return result
    return wrapper

@timer
def slow_function():
    time.sleep(1)

    print("Done")


slow_function()


print("-" * 40)


@timer
@decorator
def greet():
    print("Hello!")


greet()
```
