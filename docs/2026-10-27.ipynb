```python
def solution(triangle: list[list[int]]) -> int:
    dp = [row[:] for row in triangle]  # 값 복사해서 시작

    for i in range(1, len(dp)):
        for j in range(i + 1):
            left = dp[i-1][j-1] if j > 0 else 0
            right = dp[i-1][j] if j < i else 0
            dp[i][j] += max(left, right)
    return max(dp[-1])


def solution2(triangle: list[list[int]]) -> int:
    row = len(triangle)  # 줄 수 -> 5
    col = len(triangle[-1])  # 마지막 줄의 칸 수 -> 5

    rect = [[0] * col for _ in range(row)]  # 사각형으로 구성

    for i in range(row):
        for j in range(len(triangle[i])):
            if i == 0 and j == 0:
                rect[i][j] = triangle[i][j]
                continue
            if j == 0:
                rect[i][j] = rect[i-1][j] + triangle[i][j]
            else:
                rect[i][j] = max(rect[i-1][j], rect[i-1][j-1]) + triangle[i][j]


print(solution([[7], [3, 8], [8, 1, 0], [2, 7, 4, 4], [4, 5, 2, 6, 5]]))  # 30
print(solution2([[7], [3, 8], [8, 1, 0], [2, 7, 4, 4], [4, 5, 2, 6, 5]]))  # 30
```
