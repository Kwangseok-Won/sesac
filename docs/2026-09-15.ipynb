generator
* 지연된 return (lazy return)

전체 값을 한 번에 메모리에 올리지 않고, 필요할 때마다 한 개씩 생성함
→ 메모리 효율성 극대화
return과 비슷하지만, 값을 반환한 후 함수 상태를 저장하고 중단
다음 next() 호출 시 이전 상태부터 이어서 실행
메모리 절약: 큰 데이터를 한 번에 다루지 않고 필요한 순간에 하나씩 처리

```python
def my_range(n):
    i = 0

    while i < n:
        yield i
        i += 1


for num in my_range(5):
    print(num)

print("-" * 40)

nums = range(1000000)
squares = (x ** 2 for x in nums if x % 2 == 0)
for i in squares:
    if i > 100:
        break
    print(i)
```
