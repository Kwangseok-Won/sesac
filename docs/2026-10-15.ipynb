```python
# https://school.programmers.co.kr/learn/courses/30/lessons/138476?language=python3
def solution(k: int, tangerine: list[int]) -> int:
    answer = 0

    kv = {}

    for size in tangerine:
        kv[size] = kv.get(size, 0) + 1

    sorted_items = sorted(kv.items(), key=lambda x: x[1], reverse=True)
    counts = [x[1] for x in sorted_items]
    
    for count in counts:
        k -= count
        answer += 1

        if k <=0:
            break

    return answer


def solution2(k: int, tangerine: list[int]) -> int:
    answer = 0

    kv = {}

    for size in tangerine:
        kv[size] = kv.get(size, 0) + 1

    counts = sorted(kv.values(), reverse=True)
    for count in counts:
        k -= count
        answer += 1

        if k <= 0:
            break

    return answer


print(solution2(6, [1, 3, 2, 5, 4, 5, 2, 3]))  # 3
print(solution2(4, [1, 3, 2, 5, 4, 5, 2, 3]))  # 2
print(solution2(2, [1, 1, 1, 1, 2, 2, 2, 3]))  # 1
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12951?language=python3

def jaden_case(s: str) -> str:
    answer = ""

    splitted = s.split(" ")
    capitalized = []

    for part in splitted:
        # for p in part:
        #     if p.isnumeric():
        #         break
        capitalized.append(part.capitalize())

    answer = " ".join(capitalized)
    
    return answer


def jaden_case2(s: str) -> str:
    answer = ""

    for idx in range(len(s)):
        if idx == 0 or s[idx-1] == " ":
            answer += s[idx].upper()
        else:
            answer += s[idx].lower()

    return answer


print(jaden_case("3people unFollowed me"))  # "3people Unfollowed Me"
print(jaden_case("for the last week"))  # "For The Last Week"
```
