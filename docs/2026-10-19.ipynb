```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42842?language=python3

def solution(brown: int, yellow: int) -> list[float]:
    answer: list[float] = []

    # brown + yellow == int(width) * height

    for height in range(1, int(brown/2)):
        width = (brown + yellow) / height
        if width.is_integer():
            if 2 * width + 2 * height - 4 == brown and width >= height:
                return [int(width), height]

    return answer


print(solution(10, 2))  # [4, 3]
print(solution(8, 1))  # [3, 3]
print(solution(24, 24))  # [8, 6]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12914?language=python3

def solution(n: int) -> int:
    answer = 0

    dp = [0] * (n+1)  # dp[i] = i칸까지 가는 방법의 수
    
    dp[1] = 1
    if n >= 2:
        dp[2] = 2

    for i in range(3, n + 1):
        dp[i] = (dp[i-1] + dp[i-2]) % 1234567

    return dp[n]
    
    return answer


print(solution(4))  # 5
print(solution(3))  # 3

print(solution(2))  # 2
print(solution(5))  # 8
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181886?language=python3

def solution(names: list[str]):
    answer = []

    for idx in range(len(names)):
        if idx % 5 == 0:
            answer.append(names[idx:][0])
    
    return answer


print(solution(["nami", "ahri", "jayce", "garen", "ivern", "vex", "jinx"]))  # ["nami", "vex"]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181885?language=python3

def solution(todo_list: list[str], finished: list[bool]) -> list[str]:
    answer = []

    for todo in zip(todo_list, finished):
        not_finished, is_finished = todo

        if not is_finished:
            answer.append(not_finished)

    return answer


print(solution(["problemsolving", "practiceguitar", "swim", "studygraph"], [True, False, True, False]))  # ["practiceguitar", "studygraph"]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181884?language=python3

def solution(numbers: list[int], n: int) -> int:
    answer = 0

    for number in numbers:
        answer += number
        if answer > n:
            return answer

    return answer


print(solution([34, 5, 71, 29, 100, 34], 123))  # 139
print(solution([58, 44, 27, 10, 100], 139))  # 239
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181883?language=python3

def solution(arr: list[int], queries: list[list[int]]) -> list[int]:
    answer = []

    for query in queries:
        s, e = query

        for idx in range(len(arr)):
            if s <= idx <= e:
                arr[idx] = arr[idx] + 1

    answer = arr

    return answer


print(solution([0, 1, 2, 3, 4], [[0, 1],[1, 2],[2, 3]]))  # [1, 3, 4, 4, 4]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181882?language=python3


def solution(arr: list[int]) -> list[int]:
    answer = []

    for a in arr:
        if a % 2 == 0:
            if a >= 50:
                answer.append(a / 2)
        else:
            if a < 50:
                answer.append(a * 2)

    return answer


print(solution([1, 2, 3, 100, 99, 98]))  # [2, 2, 6, 50, 99, 49]
```
