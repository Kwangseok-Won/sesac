```python
def find_longest_sentence():
    """
    가장 긴 문장 찾기

    문제 설명: 사용자로부터 여러 문장을 . 기호로 구분하여 입력받고, 그 중 가장 긴 문장을 찾아 출력하시오.
    문장 끝의 마침표(.)는 제거해서 비교하고, 가장 긴 문장은 출력할 때 양쪽 공백을 제거한 형태로 출력하시오.
    :return:
    """

    sentences = input("여러 문장을 . 기호로 구분하여 입력하시오: ")

    splitted = sentences.split(".")
    length = 0
    idx = 0

    for i in range(len(splitted)):
        if splitted[i] != ".":
            if length < len(splitted[i]):
                length = len(splitted[i])
                idx = i

    return splitted[idx]


print(find_longest_sentence())
```

```python
def run_length_encoding(orig: str) -> str:
    """
    문자열 압축 (Run-Length Encoding)
    문제 설명: 사용자로부터 문자열을 입
    력받아, 연속된 문자가 반복되는 횟수만
    큼 압축한 문자열을 출력하시오.
    같은 문자가 연속으로 여러 번 등장
    하면 문자+횟수로 표현
    한 번만 등장한 문자는 그대로 출력
    대소문자 구분함

    예시:
    입력: "aaabbcaa"
    출력: "a3b2ca2"
    :param orig:
    :return:
    """
    compressed = ""
    prev_char = ""

    for i in range(len(orig)):
        count = 1

        if orig[i] == prev_char:
            continue

        for j in range(i + 1, len(orig)):
            if orig[i] == orig[j]:
                prev_char = orig[i]
                count += 1
            else:
                break

        compressed += orig[i]

        if count != 1:
            compressed += str(count)

    return compressed


s = "aaabbcaa"
encoded = run_length_encoding(s)
print("압축 결과: ", encoded)
```

```python
def solution(prices: list[int]) -> list[int]:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/42584?language=python3
    :param prices:
    :return:
    """

    n = len(prices)
    answer = [0] * n  # 각 시점의 유지 시간
    stack: list[int] = []  # 아직 가격이 떨어지지 않은 인덱스 저장

    for i in range(n):
        # 현재 가격(prices[i])이 이전(스택 탑)보다 낮아지는 순간
        while stack and prices[stack[-1]] > prices[i]:
            # 가격이 떨어진 시점 처리
            idx = stack.pop()
            # 떨어지기까지 걸린 시간
            answer[idx] = i - idx  # 버틴 시간 = 현재 시점(i) - 들어간 시점(idx)

        # 아직 떨어지지 않았으므로 스택에 추가
        stack.append(i)

    print("stack: ", stack)
    print("answer: ", answer)

    # 끝까지 가격이 떨어지지 않은 시점들 처리
    while stack:
        idx = stack.pop()
        answer[idx] = n - 1 - idx  # 끝까지 유지된 시간

    return answer


print(solution([1, 2, 3, 2, 3]))
```

```python
def is_palindrome_stack(word: str) -> bool:
    cleaned = word.replace(" ", "").lower()

    stack: list[str] = []

    # 1. 스택에 모든 글자 push
    for char in cleaned:
        stack.append(char)

    # 2. 스택에서 하나씩 pop하며 원래 글자와 비교
    for char in cleaned:
        if char != stack.pop():
            return False
    return True


def is_palindrome(word: str) -> bool:
    cleaned = word.replace(" ", "").lower()
    return cleaned == "".join(reversed(cleaned))


word = input("단어를 입력하세요: ")
print("회문 판별 결과 (stack 방식): ", is_palindrome_stack(word))
print("회문 판별 결과: ", is_palindrome(word))
```

```python
# https://github.com/ndb796/python-for-coding-test/blob/master/5/5.py

# 재귀 함수와 반복문
# 반복문을 쓸 수도, 재귀 함수를 쓸 수도
# 상황에 따라 무엇이 더 나을지를 고민하자

# 팩토리얼 예제
def factorial_iterative(n: int) -> int:
    """
    반복적으로 구현한 n!
    :param n:
    :return:
    """

    result = 1
    for i in range(1, n+1): # 1부터 n까지의 수를 차례대로 곱하기
        result *= i
    return result


def factorial_recursive(n: int):
    """
    재귀적으로 구현한 n!
    :param n:
    :return:
    """

    # n이 1 이하인 경우 1을 반환
    if n <= 1:
        return 1

    # n! = n * (n - 1)! 를 그대로 코드로 작성하기
    return n * factorial_recursive(n - 1)


# 각각의 방식으로 구현한 n! 출력
print(f"반복적으로 구현: {factorial_iterative(5)}")
print(f"재귀적으로 구현: {factorial_recursive(5)}")
```

```python
# https://github.com/ndb796/python-for-coding-test/blob/master/5/2.py
# https://en.wikipedia.org/wiki/Double-ended_queue

# Queue
# 줄 서기로 이해하면 좋다
# 선입선출(Fifo - first in, first out)
# 새로운 요소는 맨 뒤에 추가됨
# 큐의 맨 앞에 있는 요소만 읽거나 삭제될 수 있음
# 예시:
# 이메일 전달
# 푸쉬 알림
# 쇼핑몰 주문 처리
# 콜센터

from collections import deque

# Queue의 활용
# 리스트만으로도 구현이 가능하지만 그럴
# 경우 시간복잡도가 O(n)이다
# 따라서 collections 라이브러리의
# deque로 대체하는 것을 추천
# append는 list에서의 append와 동일하게 오른쪽 끝에 하나 붙이는 거 :
# O(1)
# popleft는 가장 왼쪽 데이터 꺼내기 :
# O(1)

queue = deque()

queue.append(5)
queue.append(2)
queue.append(3)
queue.append(7)
queue.popleft()

queue.append(1)
queue.append(4)
queue.popleft()

print(queue) # 먼저 들어온 순서대로 출력 # deque([3, 7, 1, 4])
queue.reverse() # 다음 출력을 위해 역순으로 바꾸기
print(queue) # 나중에 들어온 원소부터 출력 # deque([4, 1, 7, 3])
```

```python
# 유틸리티 메서드 구현에 자주 사용되는 staticmethod

class StringUtils:
    @staticmethod
    def toCamelCase(text: str):
        words = iter(text.split("_"))
        return next(words) + "".join(w.title() for w in words)

    @staticmethod
    def to_snake_case(text: str):
        letters = ["_" + t.lower() if t.isupper() else t for t in text]
        return "".join(letters).lstrip("_")


print(StringUtils.toCamelCase("last_modified_date"))
print(StringUtils.to_snake_case("lastModifiedDate"))
```

In Python, **composition** is an object-oriented design where a class “has a” relationship by storing other class instances as attributes instead of inheriting from them. This makes code more flexible because you can often change behavior by replacing a component object.
[realpython.com](https://realpython.com/ref/glossary/composition)

---

### Understanding Composition in Python

#### Definition of Composition

Composition in Python is a design principle where a class is structured to include instances of other classes as attributes. This creates a "has a" relationship, allowing for more flexible and maintainable code. Instead of inheriting behavior from a parent class, a class can utilize the functionality of other classes by containing them.

#### Benefits of Using Composition
- __Flexibility__: You can easily change the behavior of a class by swapping out the component objects it contains.
- **Reduced Dependency**: Composition minimizes the dependencies between classes, making the code easier to maintain and understand.
- **Reusability**: Components can be reused across different classes, promoting code reuse without the complications of inheritance.


#### Example of Composition

Consider a scenario where you have a Car class that contains an Engine class. Instead of making Car inherit from Engine, you would define Car to have an Engine as an attribute:

```python
# https://pilab-textbook.fly.dev/#/python/lab/py-lab11d-composition

UP, RIGHT, DOWN, LEFT = 0, 1, 2, 3

class Player:
    def __init__(self, name: str, hp: int, armor: Armor, weapon: Weapon):
        self.name = name
        self.hp = hp
        self.armor = armor
        self.weapon = weapon
        self.survive = True

    def attacked(self, damage):
        damage = damage - self.armor.defense
        if damage < 0:
            damage = 0
        self.hp -= damage

        if self.hp <= 0:
            print("사망")
            self.survive = False
        else:
            print(f"아야... 남은피: {self.hp}")

    def attack(self, target):
        damage = self.weapon.damage
        print(f"{self.name}가 {target.name}를 때렸다")
        target.attacked(damage)
        return None

    def inventory(self):
        print(
            f"{self.name}의 인벤토리: "
            f"{self.armor.name} (방어력: {self.armor.defense}), "
            f"{self.weapon.name} (공격력: {self.weapon.damage})"
        )

    def move(self, direction=RIGHT):
        to = direction
        if to == RIGHT:
            to = "오른쪽"
        elif to == UP:
            to = "윗쪽"
        elif to == DOWN:
            to = "아랫쪽"
        elif to == LEFT:
            to = "왼쪽"
        print(f"{self.name}은(는) {to}으로 이동했다")

class Armor:
    def __init__(self, name = "누더기", defense = 1):
        self.name = name
        self.defense = defense


class Weapon:
    def __init__(self, name = "나뭇가지", damage = 20):
        self.name = name
        self.damage = damage


cheap_armor1 = Armor("누더기", 1)
cheap_weapon1 = Weapon("나뭇가지", 20)

pricy_armor = Armor("상륙장갑", 35)
pricy_weapon = Weapon("사이드와인더", 77)

player_gaeddong = Player("개똥이", hp=100, armor=pricy_armor, weapon=pricy_weapon)
player_cowdung = Player("소똥이", hp=100, armor=cheap_armor1, weapon=cheap_weapon1)

player_gaeddong.inventory()
player_cowdung.inventory()

player_gaeddong.attack(player_cowdung)
player_gaeddong.move()
```

```python
class Adder:
    def __init__(self, value):
        self.__value = value

    # getter - 속성을 읽을 때 호출됩니다
    @property
    def value(self):
        return self.__value

    # setter - 속성에 값을 대입할 때
    @value.setter
    def value(self, value):
        if value < 0:
            raise ValueError("value는 음수가 될 수 없습니다")
        self.__value = value

    def __add__(self, other):
        # 더하기 연산자가 오른쪽에 있을 때 실행
        return self.value + other

        # if isinstance(other, Adder):
        #     return self.value + other.value
        # return self.value + other

        # self.__value += other.value
        # return self

    def __radd__(self, other):
        # 더하기 연산자가 왼쪽에 있을 때 실행
        # __add__ 로 다시금 전달
        return self.__add__(other)

    def hello(self):
        print(self.__value)


one = Adder(1)
two = Adder(2)
print(one + two)

# add = one + two
# print(add.value)

three = Adder(3)
total = one + two + three
print(total)
```

```python
# 추상 클래스
from abc import ABCMeta, abstractmethod


class Character(metaclass=ABCMeta):
    def __init__(self):
        self.hp = 100
        self.attack_power = 20

    def attack(self, other, attack_kind):
        other.get_damage(self.attack_power, attack_kind)

    @abstractmethod  # Abstract Base Classes (ABCs) according to PEP 3119
    def get_damage(self, attack_power, attack_kind):
        pass


class Player(Character):
    """
    overriding
    """
    def get_damage(self, attack_power, attack_kind):
        self.hp -= attack_power


class Player2(Character):
    pass


char1 = Player()
char2 = Player2()
```
