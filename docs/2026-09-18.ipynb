```python
def solution(code: str) -> str:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181932
    :param code:
    :return:
    """
    answer = ''

    mode = 0
    ret = ""

    for idx, ch in enumerate(code):
        if mode:
            if code[idx] == "1":
                mode = 0
            else:
                if idx % 2 != 0:
                    ret += code[idx]
        else:
            if code[idx] == "1":
                mode = 1
            else:
                if idx % 2 == 0:
                    ret += code[idx]

    if not ret:
        return "EMPTY"

    answer = ret

    return answer

print(solution("abc1abc1abc"))
```

```python
def solution(ineq: str, eq: str, n: int, m: int) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181934
    :param ineq: "<"와 ">"중 하나
    :param eq: "="와 "!"중 하나
    :return: n과 m이 ineq와 eq의 조건에 맞으면 1을 아니면 0
    """
    answer = 0

    if eq == "!":
        answer = 1 if n != m else 0
        if ineq == ">":
            return 1 if n > m else 0
        else:
            return 1 if n < m else 0
    else:
        answer = 1 if n == m else 0
        if ineq == ">":
            return 1 if n >= m else 0
        else:
            return 1 if n <= m else 0


    return answer


print(solution("<", "=", 20, 50))
print(solution(">", "!", 41, 78))
```

```python
def solution(n: int) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181935
    :param n:
    :return:
    """
    answer = 0
    even = 0
    odd = 0

    if n % 2 == 0:
        for i in range(n + 1):
            if i % 2 == 0:
                even += i ** 2
        return even
    else:
        for i in range(n + 1):
            if i % 2 != 0:
                odd += i
        return odd

    return answer

print(solution(7))
print(solution(10))
```

```python
def solution(str1: str, str2: str) -> str:
    answer = ''

    for s1 in str1:
        for s2 in str2:
            answer += s1 + s2
            break

    return answer


print(solution("aaaaa", "bbbbb"))
```
