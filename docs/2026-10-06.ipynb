```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42586?language=python3

def solution(progresses: list[int], speeds: list[int]) -> list[int]:
    answer = []

    days = []
    for p, s in zip(progresses, speeds):
        print(p, s)

        day = (100 - p) // s
        if (100 - p) % s != 0:
            day += 1
        days.append(day)

    print(days)

    for i in range(len(days)):
        if days[i] == 0:
            continue
        count = 1
        for j in range(i+1, len(days)):
            if days[i] >= days[j]:
                count += 1
                days[j] = 0
            else:
                break
        answer.append(count)

    return answer


print(solution([93, 30, 55], [1, 30, 5]))
print(solution([95, 90, 99, 99, 80, 99], [1, 1, 1, 1, 1, 1]))
```

```python
def add(v, w):
    assert len(v) == len(w), "vectors must ve the same length"
    return [v_i + w_i for v_i, w_i in zip(v, w)]

assert add([1, 2, 3], [4, 5, 6]) == [5, 7, 9]

v = [1, 2, 3]
w = [4, 5, 6]
add(v, w)


def plot(v, w):
    from matplotlib import pyplot as plt
    plt.xlim(0, 10)
    plt.ylim(0, 10)
    plt.scatter(v, w)
    plt.show()


plot(v, w)
```

```python
def scalar_multiply(c: float, v):
    """Multiplies every element by c"""
    return [c * v_i for v_i in v]

assert scalar_multiply(2, [1, 2, 3]) == [2, 4, 6]
print(scalar_multiply(2, [1, 2, 3]))


vector4 = np.array([1, 2, 3])
print(vector4 * 2)
```

```python
def vector_mean(vectors):
    """Computes the element-wise average"""
    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))

assert vector_mean([[1, 2], [3, 4], [5, 6]]) == [3, 4]


vector1 = np.array([1, 2])
vector2 = np.array([3, 4])
vector3 = np.array([5, 6])
print(sum([vector1, vector2, vector3]) / 3)
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181924

def solution(arr: list[int], queries: list[list[int]]) -> list[int]:
    answer = []

    for query in queries:
        i, j = query
        arr[i], arr[j] = arr[j], arr[i]
        answer = arr

    return answer


print(solution([0, 1, 2, 3, 4], [[0, 3], [1, 2], [1, 4]]))  # [3, 4, 1, 0, 2]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181923

def solution(arr: list[int], queries: list[list[int]]) -> list[int]:
    """
    특정 쿼리의 답이 존재하지 않으면 -1을 저장합니다.
    :param arr:
    :param queries:
    :return:
    """
    answer = []
    compared: list[int] = []

    for query in queries:
        start, end, key = query

        for i in range(len(arr)):
            if start <= i <= end:
                if arr[i] > key:
                    compared.append(arr[i])

                if len(compared) >= 2:
                    answer.append(min(compared))

                compared = []


    return answer


print(solution([0, 1, 2, 4, 3], [[0, 4, 2],[0, 3, 2],[0, 2, 2]]))  # [3, 4, -1]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181921

def solution(l: int, r: int) -> list[int]:
    answer = []

    for i in range(r + 1):
        if l <= i+1 and i+1 % 5 == 0:
            answer.append(i+1)



    return answer


print(solution(5, 555))  # [5, 50, 55, 500, 505, 550, 555]
print(solution(10, 20))  # [-1]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181920

def solution(start_num: int, end_num: int) -> list[int]:
    answer = []

    for i in range(end_num + 1):
        if start_num <= i:
            answer.append(i)

    return answer


print(solution(3, 10))
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181918

def solution(arr: list[int]) -> list[int]:
    stk = []

    i = 0
    while i < len(arr):
        if len(stk) <= 0:
            stk.append(arr[i])
            i += 1
        else:
            if stk[-1] < arr[i]:
                stk.append(arr[i])
                i += 1
            else:
                stk.pop(-1)

    return stk


print(solution([1, 4, 2, 5, 3]))  # [1, 2, 3]
```
