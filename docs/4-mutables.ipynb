_list_ 관련


```python
my_list = [0, 'a', 'b']
my_list.append(1)
print("append: ", my_list)

index = 2
my_list.insert(1, 'insert')
my_list.insert(2, index)
print("insert: ", my_list)

del my_list[index]
print("del: ", my_list)

my_list.insert(2, index)
my_list.remove(2)
print("remove: ", my_list)

my_list.pop() # last
print("pop(): ", my_list)
my_list.pop(0) # first
print("pop(0): ", my_list)

my_list.extend(['extend'])
print("extend: ", my_list)
my_list += ['new_list']
print("+= : ", my_list)

copied = my_list.copy()
print("copy: ", id(copied))
copied = list(my_list)
print("list(): ", id(copied))
copied = my_list[:]
print("[:] -- ", id(copied))

print("index: ", my_list.index("new_list"))
print("in: ", "test_value" in my_list)

my_list.append("append")
my_list.append("append")
print("count: ", my_list.count("append"))

print(my_list)
print(", ".join(my_list))
print(my_list)

print("len: ", len(my_list))
print("max: ", max(my_list))
print("min: ", min(my_list))
```

```python
# tuple은 불변객체임
print(min((38, 21, 53, 62, 19)))
print(max((38, 21, 53, 62, 19)))
print(sum((38, 21, 53, 62, 19)))

names = ['John', 'Alice', 'Lisa']
for index, name in enumerate(names): # enumerate()로 감싸면, 열거 인덱스가 부여됨
    print(f"{index}: {name}")

print(enumerate(names)) # <enumerate object at 0x주소값>
```

```python
# comprehension
print([n ** 2 for n in range(10) if n % 3 == 0])

# 구구단
print([i * j
       for j in range(2, 10)
       for i in range(1, 10)])

# tuple comprehension
print(tuple(i for i in range(10)))
```
