```python
def solution2(want: list[str], number: list[int], discount: list[str]) -> int:
    answer = 0
    target = dict(zip(want, number))

    for i in range(len(discount) - 9):
        count: dict[str, int] = {}
        for item in discount[i:i+10]:
            count[item] = count.get(item, 0) + 1
        if count == target:
            answer += 1

    return answer


print(
    solution2(
        ["banana", "apple", "rice", "pork", "pot"],
        [3, 2, 2, 2, 1],
        [
            "chicken",
            "apple",
            "apple",
            "banana",
            "rice",
            "apple",
            "pork",
            "banana",
            "pork",
            "rice",
            "pot",
            "banana",
            "apple",
            "banana",
        ],
    )
)  # 3
print(
    solution2(
        ["apple"],
        [10],
        [
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
        ],
    )
)  # 0
```

```python
def solution(want: list[str], number: list[int], discount: list[str]) -> int:
    answer = 0

    # 원하는 제품을 수량만큼 펼쳐서 정렬 (한 번만 하면 됨)
    target: list[str] = []
    for item, num in zip(want, number):
        target += [item] * num
    # "banana" * 3 -> ["banana", "banana", "banana"]
    target.sort()

    for idx in range(len(discount)):
        if sorted(discount[idx : idx+10]) == target:
            answer += 1

    return answer


print(
    solution(
        ["banana", "apple", "rice", "pork", "pot"],
        [3, 2, 2, 2, 1],
        [
            "chicken",
            "apple",
            "apple",
            "banana",
            "rice",
            "apple",
            "pork",
            "banana",
            "pork",
            "rice",
            "pot",
            "banana",
            "apple",
            "banana",
        ],
    )
)  # 3
print(
    solution(
        ["apple"],
        [10],
        [
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
            "banana",
        ],
    )
)  # 0
```
