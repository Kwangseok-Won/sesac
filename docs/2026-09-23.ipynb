```python
# 스택

def is_valid_brackets(s: str) -> bool:
    stack: list[str] | list[None] = []
    pair = {
        ")": "(",
        "}": "{",
        "]": "["
    }

    for char in s:
        if char in "({[":
            stack.append(char)
        elif char in ")}]":
            if not stack or stack[-1] != pair[char]:
                return False
            stack.pop()
    return not stack


def runner():
    expr = input("괄호 문자열을 입력하세요: ")
    print("입력한 괄호 검사 결과: ", is_valid_brackets(expr))


runner()
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12906?language=python3

def solution1(arr: list[int]) -> list[int]:
    answer: list[int] | list[None] = []

    for array_element in arr:
        if not answer or answer[-1] != array_element:
            answer.append(array_element)

    return answer


def solution2(arr: list[int]) -> list[int]:
    answer: list[int] | list[None] = []

    for array_index in range(len(arr)):
        if array_index == 0 or answer[-1] != arr[array_index]:
            answer.append(arr[array_index])

    return answer


print(solution1([1, 1, 3, 3, 0, 1, 1]))
print(solution1([4, 4, 4, 3, 3]))
print("-" * 40)
print(solution2([1, 1, 3, 3, 0, 1, 1]))
print(solution2([4, 4, 4, 3, 3]))
```

```python
def solution(array: list[int], commands: list[list[int]]) -> list[int]:
    """
    # https://school.programmers.co.kr/learn/courses/30/lessons/42748?language=python3

    :param array:
    :param commands: [[i, j, k]]
    :return:
    """
    answer = []
    i, j, k = 0, 0, 0

    for command in commands:
        i, j, k = command
        sliced = array[i-1 : j]
        sorted_array = sorted(sliced)
        answer.append(sorted_array[k - 1])


    return answer

def solution2(array: list[int], commands: list[list[int]]) -> list[int]:
    answer = []

    for command in commands:
        i, j, k = command
        arr = array[i-1 : j]
        arr.sort()
        answer.append(arr[k - 1])


    return answer


print(solution([1, 5, 2, 6, 3, 7, 4], [[2, 5, 3], [4, 4, 1], [1, 7, 3]]))
print(solution2([1, 5, 2, 6, 3, 7, 4], [[2, 5, 3], [4, 4, 1], [1, 7, 3]]))
```

```python
def solution(num: int, total: int) -> list[int]:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/120923?language=python3
    :param num:
    :param total:
    :return:
    """
    answer = []

    num_sum = sum(range(num))

    start = int((total - num_sum) / num)
    answer = list(range(start, start + num))

    return answer

print(solution(3, 12))
print(solution(5, 15))
print(solution(4, 14))
print(solution(5, 5))
```

```python
# https://docs.python.org/3.14/library/zlib.html

import zlib

s = b"which " * 100
print("crc checksum: ", zlib.crc32(s))

print("before: ", len(s))
t = zlib.compress(s)
print("after: ", len(t))

decompressed = zlib.decompress(t)
print("restored: ", len(decompressed))

print("-" * 40)
print("compressed: ", t)
print("decompressed: ", decompressed)


print("-" * 40)
# 압축률 확인
# 작은 데이터는 오히려 커질 수도 있다
samples = [
    b"abc",  # 3 ->   11
    b"hello world", # 11 ->   19
    b"witch which has which witches wrist watch",
    b"which " * 100,
]
for s in samples:
    c = zlib.compress(s)
    print(
        f"{len(s):4d} -> {len(c):4d} "
        f"({len(c)/len(s):.2f}x)"
    )
```
