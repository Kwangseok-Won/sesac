```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    "Favorite Food": [
        "Alfredo Pasta",
        "Bobba",
        "Croissants",
        "Danish Rolls",
        "Edaame",
        "Fish",
        "Ice Cream"
    ]
},
    index=pd.Index(["Alfred", "Bob", "Cris", "Daniel", "Edward", "Fitz", "Ivan"],
                   name="Name"),
)
df2 = pd.DataFrame({
    "Favorite Country": [
        "Albania",
        "Croatia",
        "Denmark",
        "Fiji",
        "Germany",
        "Haiti"
    ]},
    index=pd.Index(["Alfred", "Cris", "Daniel", "Fitz", "Gerald", "Hayden"],
                   name="Name"),
)
display(df1)
display(df2)
print("-" * 40)

```

```python

print("outer join - 합집합")
display(df1.join(df2, how="outer"))

print("inner join - 교집합")
display(df1.join(df2, how="inner"))

print("기본값 - left join")
display(df1.join(df2))

print("right join")
display(df1.join(df2, how="right"))
```

```python
# merge - join 기능으로부터 더욱 보강한 것

# merge를 사용하기 전엔 reset_index를 먼저 실행해야 한다
df1 = df1.reset_index()
df2 = df2.reset_index()
display(df1)
display(df2)

# 교집합(inner)가 기본값. how=에다가 "outer", "left", "right" 선택 가능
display(pd.merge(df1, df2))

display(df1.merge(df2))

df3 = pd.DataFrame({"Favorite Course": ["Criminomogy", "Design", "Fitness"],
                    "Favorite Country": ["Croatia", "Denmark", "Fiji"]})

display(pd.merge(df2, df3))
```

```python
df4 = pd.DataFrame({"Name": ["Cris", "Daniel", "Fitz", "Peter"],
                    "Favorite Country": ["Croatia", "Denmark", "Fiji", "Croatia"],
                    "Favorite Course": ["Criminomogy", "Design", "Fitness", "Yoga"]})

display(df2)
display(df4)

print("on은 기준컬럼")
display(df2.merge(df4, on=["Favorite Country"]))
```

```python
print("stack / unstack")

index = pd.MultiIndex.from_tuples([("bar", "one"), ("bar", "two"),
                                   ("baz", "one"), ("baz", "two")],
                                  names=["class", "name"])
df2 = pd.DataFrame([[1, 2], [3, 4], [5, 6], [7, 8]], index=index, columns=["eng", "math"])
display(df2)

stacked = df2.stack()
display(stacked)

print("stacked.index.names: ", stacked.index.names)

print("unstack")
display(stacked.unstack())

print(stacked.unstack().index.names)
display(stacked.unstack("name"))  # 또는 stacked.unstack(1)
display(stacked.unstack("class"))  # 또는 stacked.unstack(0)
```

```python
class Pet:
    all = []

    def __init__(self, name):
        self.name = name
        self.hunger = 5
        Pet.all.append(self)

    def feed(self):
        self.hunger -= 1

    @classmethod
    def to_df(cls):
        return pd.DataFrame({"이름": p.name, "배고픔": p.hunger} for p in cls.all)
```
