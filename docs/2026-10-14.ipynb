```python
# https://school.programmers.co.kr/learn/courses/30/lessons/131705?language=python3

def solution(number: list[int]) -> int:
    answer = 0
    n = len(number)

    for i in range(n):
        for j in range(i+1, n):
            for k in range(j+1, n):
                if number[i] + number[j] + number[k] == 0:
                    answer += 1
    
    return answer


print(solution([-2, 3, 0, 2, -5]))  # 2
print(solution([-3, -2, -1, 0, 1, 2, 3]))  # 5
print(solution([-1, 1, -1, 1]))  # 0
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/86491?language=python3

def solution(sizes: list[list[int]]) -> int:
    answer = 0

    for size in sizes:
        size.sort()
    
    max_w = max([s[0] for s in sizes])
    max_h = max([s[1] for s in sizes])

    answer = max_w * max_h

    return answer


print(solution([[60, 50], [30, 70], [60, 30], [80, 40]]))  # 4000
print(solution([[10, 7], [12, 3], [8, 15], [14, 7], [5, 15]]))  # 120
print(solution([[14, 4], [19, 6], [6, 16], [18, 7], [7, 11]]))  # 133
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181894?language=python3

def solution(arr: list[int]) -> list[int]:
    answer = []

    indices = []

    if 2 in arr:
        for idx in range(len(arr)):
            if arr[idx] == 2:
                indices.append(idx)
    else:
        return [-1]

    start = min(indices)
    end = max(indices)
    answer = arr[start : end + 1]
    
    return answer


print(solution([1, 2, 1, 4, 5, 2, 9]))  # [2, 1, 4, 5, 2]
print(solution([1, 2, 1]))  # [2]
print(solution([1, 1, 1]))  # [-1]
print(solution([1, 2, 1, 2, 1, 10, 2, 1]))  # [2, 1, 2, 1, 10, 2]
```
