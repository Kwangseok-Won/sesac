```python
# https://school.programmers.co.kr/learn/courses/30/lessons/155652?language=python3

def solution(s: str, skip: str, index: int) -> str:
    answer = ""
    all_list = []

    # skip에 포함되지 않는 알파벳만 all_list에 추가
    for ch in range(97, 97 + 26):
        if chr(ch) in skip:
            continue
        all_list.append(chr(ch))

    list_len = len(all_list)

    for i in range(len(s)):
        idx_all = all_list.index(s[i])

        # 인덱스가 리스트 길이를 넘어가면 나머지 연산으로 처리
        answer += all_list[(idx_all + index) % list_len]

    return answer


print(solution("aukks", "wbqd", 5))  # happy
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181912?language=python3

def solution(intStrs: list[str], k: int, s: int, l: int) -> list[int]:
    answer = []

    for int_str in intStrs:
        end = len(int_str)
        parted = int_str[s : end]

        if s+l < len(int_str):
            parted = int_str[s:end]

        print(parted)

        if k < int(parted):
            answer.append(int(parted))
    
    
    return answer


print(
    solution(["0123456789", "9876543210", "9999999999999"], 50000, 5, 5)
)  # [56789, 99999]
```

```python
solution = lambda intStrs, k, s, l: [
    int(intStrs[i][s : s + l])
    for i in range(len(intStrs))
    if int(intStrs[i][s : s + l]) > k
]


print(
    solution(["0123456789", "9876543210", "9999999999999"], 50000, 5, 5)
)  # [56789, 99999]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181911?language=python3

def solution(my_strings: list[str], parts: list[list[int]]):
    answer = ""

    for idx_my_string in range(len(my_strings)):
        for idx_part in range(len(parts)):
            if idx_my_string == idx_part:
                s, e = parts[idx_part]
                answer += my_strings[idx_my_string][s : e + 1]

    return answer


print(
    solution(
        ["progressive", "hamburger", "hammer", "ahocorasick"],
        [[0, 4], [1, 2], [3, 5], [7, 7]],
    )
)  # "programmers"
```
