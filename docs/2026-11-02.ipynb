```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42584?language=python3

def solution(prices: list[int]) -> list[int]:
    n = len(prices)
    answer = [0] * n

    for i in range(n):
        for j in range(i + 1, n):
            answer[i] += 1  # 1초 지났다고 카운트

            # 나보다 가격이 떨어지면 반복문 중단
            if prices[i] > prices[j]:
                break

    return answer


def solution2(prices: list[int]):
    n = len(prices)
    answer = [0] * n
    stack = []  # (시점, 가격)

    for idx, p in enumerate(prices):
        while stack and stack[-1][1] > p:
            j, _ = stack.pop()
            answer[j] = idx - j
        stack.append((idx, p))

    for j, _ in stack:
        answer[j] = n - 1 - j

    return answer


# O^2
def solution3(prices):
    answer = []
    for i in range(len(prices)):
        count = 0
        for j in range(i+1, len(prices)):
            count += 1
            if prices[i] > prices[j]:
                break
        answer.append(count)
    return answer


print(solution([1, 2, 3, 2, 3]))  # [4, 3, 1, 1, 0]
print(solution([1, 1, 1, 1, 1]))  # [4, 3, 2, 1, 0]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42840
from math import ceil


def solution(answers: list[int]) -> list[int]:
    answer = []

    scores = []
    members = [
        [1, 2, 3, 4, 5],
        [2, 1, 2, 3, 2, 4, 2, 5],
        [3, 3, 1, 1, 2, 2, 4, 4, 5, 5],
    ]

    for mem in members:
        temp = ceil(len(answers) / len(mem)) * mem
        scores.append(sum(a == t for a, t in zip(answers, temp)))

    best = max(scores)
    for idx, score in enumerate(scores):
        if score == best:
            answer.append(idx + 1)
    
    return answer


print(solution([1, 2, 3, 4, 5]))  # [1]
print(solution([1, 3, 2, 4, 2]))  # [1, 2, 3]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42839
from itertools import permutations


def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def solution(numbers: str) -> int:
    answer = 0
    num = []

    for idx in range(len(numbers)):
        num.append(numbers[idx])

    unique_numbers = set()
    for i in range(1, len(num) + 1):
        for p in permutations(num, i):
            n = int(''.join(p))
            if is_prime(n):
                unique_numbers.add(n)
    answer = len(unique_numbers)

    return answer


def solution_2(numbers: str) -> int:
    nums = set()
    for n in range(1, len(numbers) + 1):
        for temp in permutations(numbers, n):
            nums.add(int("".join(temp)))

    answer = 0
    for n in nums:
        if is_prime(n):
            answer += 1


print(solution("17"))  # 3
print(solution("011"))  # 2

print(solution_2("17"))  # 3
print(solution_2("011"))  # 2
```

```python
from itertools import permutations


def solution(digits):
    best = -1

    for a, b, c in permutations(digits, 3):
        num = a * 100 + b * 10 + c
        if num % 2 == 0:
            best = max(best, num)

    return best


print(solution(digits="17"))
```
