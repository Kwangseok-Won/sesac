```python
def solution(storage: list[str], num: list[int]) -> str:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/250126
    :param storage:
    :param num:
    :return:
    """
    clean_storage: list[str] = []
    clean_num: list[int] = []
    for i in range(len(storage)):
        if storage[i] in clean_storage:
            pos = clean_storage.index(storage[i])
            clean_num[pos] += num[i]
        else:
            clean_storage.append(num[i])
            clean_num.append(num[i])

    # 아래 코드에는 틀린 부분이 없습니다.

    max_num = max(clean_num)
    answer = clean_storage[clean_num.index(max_num)]
    return answer

print(solution(storage=["pencil", "pencil", "pencil", "book"], num=[2, 4, 3, 1]))
print(solution(storage=["doll", "doll", "doll", "doll"], num=[1, 1, 1, 1]))
print(solution(storage=["apple", "steel", "leaf", "apple", "leaf"], num=[5, 3, 5, 3, 7]))
print(solution(storage=["mirror", "net", "mirror", "net", "bottle"], num=[4, 1, 4, 1, 5]))
```

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def solution(numer1, denom1, numer2, denom2):
    answer = [numer1 * denom2 + numer2 * denom1, denom1 * denom2]
    GCD = gcd(answer[0], answer[1])
    answer[0], answer[1] = answer[0] // GCD, answer[1] // GCD
    return answer
```

```python
def solution(numbers: list[int], our_score: list[int], score_list: list[int]) -> list[str]:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/250128
    :param numbers:
    :param our_score:
    :param score_list:
    :return:
    """
    answer = []
    for i in range(len(numbers)):
        if our_score[i] == score_list[numbers[i] - 1]:
            answer.append("Same")
        else:
            answer.append("Different")

    return answer

print(solution(numbers=[1], our_score=[100], score_list=[100, 80, 90, 84, 20]))
print(solution(numbers=[3, 4], our_score=[85, 93], score_list=[85, 92, 38, 93, 48, 85, 92, 56]))
```

```python
def solution(my_string: str, overwrite_string: str, s: int) -> str:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181943
    :param my_string: 숫자와 알파벳
    :param overwrite_string: 숫자와 알파벳
    :param s: 덮어쓰려는 곳의 시작인덱스
    :return: 문자열 my_string의 인덱스 s부터 overwrite_string의 길이만큼을 문자열 overwrite_string으로 바꾼 문자열
    """

    # my_string.repl
    to_replace = s + len(overwrite_string)
    start = my_string[:s]

    replaced = my_string[s:to_replace]
    replaced = overwrite_string

    end = my_string[to_replace:]

    answer = f"{start}{replaced}{end}"
    return answer

print(solution(my_string="He11oWor1d", overwrite_string="lloWorl", s=2))
print(solution(my_string="Program29b8UYP", overwrite_string="merS123", s=7))
```

```python
def solution(nickname: str) -> str:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/340200
    :param nickname: 닉네임의 길이가 4자 이상 8자 이하여야합니다.
    수정된 닉네임의 길이가 4 미만일 경우 뒤에 소문자 o를 길이가 4가 될때까지 이어붙입니다.
    :return:
    닉네임 "WORLDworld"는 1, 2, 3, 4, 6 단계를 거쳐 "VV0RLDvv"가 됩니다.
    "WORLDworld" -> "WORLDworId" -> "WORLDvvorId" -> "VVORLDvvorId" -> "VV0RLDvvorId" -> "VV0RLDvv"
    닉네임 "GO"는 4, 5 단계를 거쳐 "G0oo"가 됩니다.
    "GO" -> "G0" -> "G0oo"
    """
    answer = ""
    for letter in nickname:
        if letter == "l":
            answer += "I"
        elif letter == "w":
            answer += "vv"
        elif letter == "W":
            answer += "VV"
        elif letter == "O":
            answer += "0"
        else:
            answer += letter

    # if len(answer) < 3:
    #     answer += "o" * (4 - len(answer))
    while len(answer) < 4:
        answer += "o"

    if len(answer) > 8:
        answer = answer[:8]
    return answer

print(solution("WORLDworld"))
print(solution("GO"))
```

```python
def func1(num: int) -> int:
    if 0 > num:
        return 0
    else:
        return num


def func2(num: int):
    if num > 0:
        return 0
    else:
        return num


def func3(station: list[str]):
    num = 0
    for people in station:
        if people == "Off":
            num += 1
    return num


def func4(station: list[str]):
    num = 0
    for people in station:
        if people == "On":
            num += 1
    return num


def solution(seat: int, passengers: list[list[str]]) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/340201
    :param seat: 버스의 좌석 개수
    :param passengers: 기점에서 출발한 버스가 순서대로 방문한 정거장에서 승객이 승/하차한 정보를 담은 2차원 문자열 리스트
    :return: 버스에 남아있는 좌석의 개수
    """

    num_passenger = 0
    for station in passengers:
        num_passenger += func4(station)
        num_passenger -= func3(station)
    answer = func1(seat - num_passenger)
    return answer

print(solution(seat=5, passengers=[
    ["On", "On", "On"], # (0명 하차, 3명 승차) -> 2
    ["Off", "On", "-"], # (1명 하차, 1명 승차) -> 2
    ["Off", "-", "-"] # (1명 하차, 0명 승차) -> 3
])) # 3
print(solution(seat=10, passengers=[
    ["On", "On", "On", "On", "On", "On", "On", "On", "-", "-"],
    ["On", "On", "Off", "Off", "Off", "On", "On", "-", "-", "-"],
    ["On", "On", "On", "Off", "On", "On", "On", "Off", "Off", "Off"],
    ["On", "On", "Off", "-", "-", "-", "-", "-", "-", "-"]
])) # 0
```

```python
def solution(storage: int, usage: int, change: list[int]) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/340202
    :param storage: 현재 저수지에 저장된 물의 양을 나타내는 정수
    :param usage: 지난 달 물 사용량을 나타내는 정수
    :param change: 월별 물 사용량이 전 달 대비 어떻게 변하는지 저장된 정수 리스트
    :return: 몇 달 뒤 물이 부족해지는지 return. 가뭄이 끝날때 까지 저수지의 물이 남아 있다면 -1을 return
    """
    total_usage = 0
    for i in range(len(change)):
        # usage = total_usage * change[i] / 100
        usage += usage * change[i] / 100
        total_usage += usage
        if total_usage > storage:
            return i

    return -1


print(solution(storage=5141, usage=500, change=[10, -10, 10, -10, 10, -10, 10, -10, 10, -10]))  # -1
print(solution(storage=1000, usage=2000, change=[-10, 25, -33]))  # 0
```

```python
def solution(cpr):
    """ https://school.programmers.co.kr/learn/courses/30/lessons/340203 """
    answer = []
    basic_order = ["check", "call", "pressure", "respiration", "repeat"]
    for action in cpr:
        for i in range(len(cpr)):
            if action == basic_order[i]:
                answer.append(i + 1)
    return answer


print(solution(["call", "respiration", "repeat", "check", "pressure"]))
print(solution(["respiration", "repeat", "check", "pressure", "call"]))
```
