```python
def make_counter():
    count = 0

    def counter():
        nonlocal count #중첩함수 내부에서 바깥(Enclosing) 함수의 변수를 수정할 때 사용 # nonlocal 없이는 바깥 변수는 읽기만 가능하고, 쓰면 새로운 지역변수가 생김 # nonlocal은 LEGB의 E(Enclosing) 범위의 바인딩을 명시적으로 수정함
        count += 1
        return count

    return counter

c1 = make_counter()
print(c1())
print(c1())
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Start") # pre-hook
        result = func(*args, **kwargs)
        print("End") # post-hook

        return result
    return wrapper

@decorator
def add(x, y):
    return x + y

print(add(3, 5))
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time() # pre-hook
        result = func(*args, **kwargs)
        end = time.time() # post-hook

        print(f"실행 시간: {end - start:.4f}초")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(1)
    print("Done")

slow_function()
```

```python
import dis
import inspect
import traceback


def reverse(s: str) -> str:
    depth = len(inspect.stack()) - 1
    print("  " * depth + f"ENTER {repr(s)}")

    # traceback.print_stack(limit=20)

    if s == "":
        print("  " * depth + "RETURN ''")
        return ""

    result = reverse(s[1:]) + s[0]

    print("  " * depth + f"RETURN {repr(result)}")

    return result

print(reverse("abc"))
dis.dis(reverse)
```

```python
# https://en.wikipedia.org/wiki/Maze-solving_algorithm
maze = [
    [0, 0, 1, 0],
    [1, 0, 1, 0],
    [0, 0, 0, 0],
    [0, 1, 1, 0],
]

visited: set[int, int] = set()


def solve(x: int, y: int):
    # 범위 밖
    if x < 0 or y < 0 or x >= 4 or y >= 4:
        return False

    # 벽
    if maze[y][x] == 1:
        return False

    # 이미 방문
    if (x, y) in visited:
        return False

    # 도착
    if (x, y) == (3, 3):
        print("도착!")
        return True

    visited.add((x, y))

    return (
        solve(x + 1, y)
        or solve(x - 1, y)
        or solve(x, y + 1)
        or solve(x, y - 1)
    )


solve(0, 0)
```

```python
# 순열

n = 3
# ├── 0
# │    ├── 1
# │    │    └── 2
# │    └── 2
# │         └── 1
# │
# ├── 1
# │    ├── 0
# │    └── 2
# │
# └── 2
#      ├── 0
#      └── 1

used = [False] * n
path = []


def perm():
    if len(path) == n:
        print(path)
        return

    for i in range(n):
        if used[i]:
            continue

        used[i] = True          # 선택
        path.append(i)

        perm()                  # 재귀

        path.pop()              # 원상복구
        used[i] = False


perm()
```

```python
# 괄호 생성(Generate Parentheses)

# 가지치기(pruning) - 닫는 괄호는 항상 여는 괄호보다 적게 등장해야 한다.

def generate_parentheses(n):
    result = []

    def dfs(left, right, s):
        if left == n and right == n:
            result.append(s)
            return

        if left < n:
            dfs(left + 1, right, s + "(")

        if right < left:
            dfs(left, right + 1, s + ")")

    dfs(0, 0, "")

    return result


print(generate_parentheses(3))
```

```python
# https://velog.io/@jguuun/정렬-알고리즘#퀵-정렬quick-sort
# https://seongjaemoon.github.io/python/2017/12/16/pythonSort.html#4-퀵정렬
# https://ejklike.github.io/2017/03/04/sorting-algorithms-with-python.html#quick-sort-퀵-정렬

# O(N log N)
def quicksort(arr: list[int]):
    if len(arr) <= 1:
        return arr

    pivot = arr[0]

    left = [x for x in arr[1:] if x <= pivot]
    right = [x for x in arr[1:] if x > pivot]

    return quicksort(left) + [pivot] + quicksort(right)


quicksort([8, 4, 6, 2, 5, 1, 3, 7, 9])
```

```python
# https://en.wikipedia.org/wiki/Merge_sort
# O(n log n)

def merge(left, right):

    result = []

    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


def mergesort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = mergesort(arr[:mid])
    right = mergesort(arr[mid:])

    return merge(left, right)
```

```python
# N-Queen
# 백트래킹의 대표 문제
#
# 재귀의 핵심
# 놓는다
# ↓
# 다음 줄
# ↓
# 실패하면
# ↓
# 되돌린다(backtracking)

N = 8

board = [-1] * N

def safe(row: int, col: int) -> bool:
    for r in range(row):
        c = board[r]

        if c == col:
            return False

        if abs(c - col) == abs(r - row):
            return False

    return True


def queen(row: int) -> None:
    if row == N:
        print(board)
        return

    for col in range(N):
        if safe(row, col):
            board[row] = col
            queen(row + 1)
            board[row] = -1


queen(0)
```

```python
# sudoku solver

board = [
    [5,3,0,0,7,0,0,0,0],
    [6,0,0,1,9,5,0,0,0],
    [0,9,8,0,0,0,0,6,0],
    [8,0,0,0,6,0,0,0,3],
    [4,0,0,8,0,3,0,0,1],
    [7,0,0,0,2,0,0,0,6],
    [0,6,0,0,0,0,2,8,0],
    [0,0,0,4,1,9,0,0,5],
    [0,0,0,0,8,0,0,7,9],
]


def find_empty():
    """
    빈 칸 찾기
    :return:
    """
    for r in range(9):
        for c in range(9):
            if board[r][c] == 0:
                return r, c

    return None, None


def valid(row, col, num):
    """
    숫자 검사
    :param row:
    :param col:
    :param num:
    :return:
    """
    for i in range(9):
        if board[row][i] == num:
            return False

    for i in range(9):
        if board[i][col] == num:
            return False

    box_r = (row // 3) * 3
    box_c = (col // 3) * 3

    for r in range(box_r, box_r + 3):
        for c in range(box_c, box_c + 3):
            if board[r][c] == num:
                return False

    return True


def solve():
    """
    재귀
    :return:
    """
    row, col = find_empty()

    if row is None:
        return True

    for num in range(1, 10):
        if valid(row, col, num):
            # Backtracking
            board[row][col] = num
            if solve():
                return True
            board[row][col] = 0

    return False


# 실행
solve()

for row in board:
    print(row)
```

```python
def hanoi(n, src, dst, aux):
    if n == 1:
        print(f"{src} -> {dst}")
        return

    # 위의 n-1개를 치운다.
    hanoi(n - 1, src, aux, dst)

    # 가장 큰 원반 이동
    print(f"{src} -> {dst}")

    # 치워둔 n-1개를 다시 올린다.
    hanoi(n - 1, aux, dst, src)
```
