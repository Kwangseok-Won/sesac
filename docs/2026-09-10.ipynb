```python
numbers: list[int] = list(range(0, 50, 3))
print(f"numbers: {numbers}")

odd = []; even = []; multiplier = []; x = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)
        if num % 3 == 0:
            x.append(num) # 짝수이면서 3의 배수
    elif num % 3 == 0:
        x.append(num) # 홀수 중에서 3의 배수

    if num % 3 == 0:
        odd.append(num)

print(f"짝수: {even}")
print(f"홀수: {odd}")
print(f"짝수이면서\n 3의 배수: {x}")

```

# 보조제어문
* continue
* break
* pass

```python
truthy = -1
idx = 0

while truthy:
    idx += 1

    if idx <= 3:
        pass # 아래 내용 무시
    elif idx < 10:
        continue
    else:
        break

    print("here!")

print("-" * 50)

idx = 0
while truthy:
    idx += 1

    if idx < 3:
        continue
    elif idx <= 10:
        print("출력")
    else:
        print("끝")
        break
```

```python
x = "hello world"
for elem in x:
    print(elem)

    if elem == "o":
        break
else:
    print("for-else 끝")
```

```python
x = "hello world"
idx = 0

while idx < 11:
    print(x[idx])
    idx += 1

    if idx > 5:
        break
else:
    print("while-else 끝")
```

```python
a = [1, 2, 3]
b = [4, 5, 6]
ESCAPE_FLAG = False

for outer_idx in a:
    if ESCAPE_FLAG:
        break

    for inner_idx, inner_value in enumerate(b):
        print(outer_idx, inner_value)

        if inner_idx == 1:
            ESCAPE_FLAG = True # 한 번에 중첩반복문 탈출
            break
        else:
            pass

print("-" * 50)
```
