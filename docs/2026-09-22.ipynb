```python
import time

from for_dict import value


def prime_printer_1(n):
    start = time.time()
    prime_list = []
    for i in range(2, n+1):
        detector = 0
        for k in prime_list:
            if i % k == 0:
                detector += 1
        if detector == 0:
            prime_list.append(i)

    end = time.time()
    return end - start

print(f"required {prime_printer_1(100):10.12f} seconds")
print(f"required {prime_printer_1(1_000):10.12f} seconds")
print(f"required {prime_printer_1(10_000):10.12f} seconds")
print(f"required {prime_printer_1(100_000):10.12f} seconds")
```

```python
import time

def prime_printer_2(n):
    start = time.time()
    prime_list = []
    for i in range(1, n + 1):
        detector = 0
        for k in range(2, i): # n^2
            if i % k == 0:
                detector += 1
        if detector == 0:
            prime_list.append(i)

    end = time.time()
    return end - start

print(f"required {prime_printer_2(100):10.12f} seconds")
print(f"required {prime_printer_2(1_000):10.12f} seconds")
print(f"required {prime_printer_2(10_000):10.12f} seconds")
print(f"required {prime_printer_2(100_000):10.12f} seconds")
```

```python
# Logarithmic Complexity – O(log N)

# 이진탐색
def binary_search(array: list[int], target: int, left: int, right: int):
    middle_idx = (left + right) // 2
    print(middle_idx)

    middle = array[middle_idx]

    if target == middle:
        print("answer {}".format(middle_idx))
    elif middle > target:
        binary_search(array, target, left, middle_idx - 1)
    elif middle < target:
        binary_search(array, target, middle_idx + 1, right)
    else:
        return False


target = 25
m_list = [30, 94, 92, 21, 37, 25, 47, 25, 53, 98, 19, 32, 32, 7]
length = len(m_list)

m_list = sorted(m_list)
left = 0
right = length - 1

binary_search(m_list, target, 0, right)
```

```python
# https://arxiv.org/abs/2110.01111

import random
import copy

help(random.randint)

def new_sort(L):
    l = len(L)

    for i in range(l):
        for j in range(l):
            if L[i] < L[j]:
                L[i], L[j] = L[j], L[i]
    return L


my_list = [random.randint(0, 1000) for _ in range(10)]
old_list = copy.deepcopy(my_list)
sorted_out = new_sort(my_list)

for i, j in zip(old_list, sorted_out):
    print(f"{i} {j}")
```

```python
array = [8, 4, 6, 2, 9, 1, 3, 7, 5]

def selection_sort(array: list[int]):
    n = len(array)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if array[j] < array[min_index]:
                min_index = j
        array[i], array[min_index] = array[min_index], array[i]
        print(array[:i+1])


print("before: ", array)
selection_sort(array)
print("after: ", array)
```

```python
array = [8, 4, 6, 2, 9, 1, 3, 7, 5]

def insertion_sort(array: list[int]):
    n = len(array)
    for i in range(1, n):
        for j in range(i, 0, -1):
            if array[j - 1] > array[j]:
                array[j - 1], array[j] = array[j], array[j - 1]
        print(array[:i+1])


print("before: ", array)
insertion_sort(array)
print("after: ", array)
```

```python
# O(n log n)

array = [8, 4, 6, 2, 9, 1, 3, 7, 5]

def merge_sort(array: list[int]):
    if len(array) < 2:
        return array
    mid = len(array) // 2
    low_arr = merge_sort(array[:mid])
    high_arr = merge_sort(array[mid:])

    merged_arr = []
    l = h = 0
    while l < len(low_arr) and h < len(high_arr):
        if low_arr[l] < high_arr[h]:
            merged_arr.append(low_arr[l])
            l += 1
        else:
            merged_arr.append(high_arr[h])
            h += 1
    merged_arr += low_arr[l:]
    merged_arr += high_arr[h:]
    print(merged_arr)

    return merged_arr


print("before: ", array)
array = merge_sort(array)
print("after: ", array)
```

```python
# https://velog.io/@jguuun/정렬-알고리즘#퀵-정렬quick-sort
# https://seongjaemoon.github.io/python/2017/12/16/pythonSort.html#4-퀵정렬
# https://ejklike.github.io/2017/03/04/sorting-algorithms-with-python.html#quick-sort-퀵-정렬

# O(N log N)

array = [8, 4, 6, 2, 5, 1, 3, 7, 9]

def quick_sort(array: list[int]):
    if len(array) <= 1:
        return array
    pivot = len(array) // 2
    front_arr, pivot_arr, back_arr = [], [], []
    for value in array:
        if value < array[pivot]:
            front_arr.append(value)
        elif value > array[pivot]:
            back_arr.append(value)
        else:
            pivot_arr.append(value)
    print(front_arr, pivot_arr, back_arr)

    return quick_sort(front_arr) + quick_sort(pivot_arr) + quick_sort(back_arr)


print("before: ", array)
array = quick_sort(array)
print("after: ", array)
```
