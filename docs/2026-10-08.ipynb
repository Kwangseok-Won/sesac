```python
def func(x: int) -> float:
    """
        docstring 요약

        Args:
            x (int): 정수 입력 값

        Returns:
            float: 소수 출력 값

        Examples:
            y = func(1)
            print(y, type(y))
    """
    
    x= float(x)

    return x


y = func("1")
print(y, type(y))

help(func)
```

```python
class Person:
    population = 0
    
    def __init__(self, name, age):
        # print(self, id(self), type(self))
        
        self.name = name
        self.age = age
        Person.population += 1

    @classmethod
    def init_from_dict(cls, d):
        return cls(d["name"], d["age"])

    @classmethod
    def count_population(cls):
        print(cls, id(cls), type(cls))
        return cls.population

    def introduce_myself(self):
        print(f"Hello, my name is {self.name}; age is {self.age}")

    @staticmethod
    def say_hi():
        print("hi!")


mr_k = Person("Mr. K", 0)
print(Person.count_population())

jill = Person.init_from_dict({"name": "Jill", "age": 10})
print(Person.count_population())

jack = Person("Jack", 20)
print(Person.count_population())

print("-" * 40)
mr_k.introduce_myself()
jill.introduce_myself()

print("-" * 40)
mr_k.say_hi()

print("-" * 40)

class Character(Person):
    def __init__(self, name, age, weapon = None):
        super().__init__(name, age)
        self.hp = 10
        self.weapon = weapon
        self.survive = True

    def set_weapon(self, weapon):
        self.weapon = weapon

    def attack(self, target):
        if self.weapon:
            damage = self.weapon.damage
        else:
            damage = 1
        target.damaged(damage)

    def damaged(self, damage):
        if self.survive:
            self.hp -= damage
            print(f"{damage} 받음 --> 남은 피: {self.hp}")

            if self.hp <= 0:
                self.survive = False
                print("죽음")
        else:
            print("이미 죽었습니다")


class Weapon:
    def __init__(self, name, damage):
        self.name = name
        self.damage = damage


stick = Weapon("몽둥이", 5)
inherited_character = Character("Person에게 상속 받은 자", 40, stick)
first_user = Character("첫 번째 유저", 20, stick)

inherited_character.say_hi()
first_user.say_hi()

inherited_character.damaged(3)
inherited_character.damaged(3)
inherited_character.damaged(3)
inherited_character.damaged(3)
inherited_character.damaged(3)
```
