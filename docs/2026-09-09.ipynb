# list

```python
x = list("hello world")
print(x, type(x), len(x))  # ['h', 'e', 'l', 'l', 'o', ' ', 'w', 'o', 'r', 'l', 'd'] <class 'list'> 11
print("".join(x))  # hello world

x = str(x)
print(x, type(x), len(x))  # ['h', 'e', 'l', 'l', 'o', ' ', 'w', 'o', 'r', 'l', 'd'] <class 'str'> 55
print(x[0])
print(x[1])
print(x[2])
```

```python
# 'reverse', 'sort'

x = [3, 1, 2]
print(x, id(x))

x.reverse()  # 가변
# x.sort(reverse=True)
print(x, id(x))

x.sort()  # 가변
print(x, id(x))

x = [1, 2, 3]
y = x
y[0] = 99
print(x, id(x))
print(y, id(y))

help(x)
```

```python
help(list.sort)
```

```python
x = 'hello world'
y = x.upper()
print(x)  # x는 불변
print(y)
```

list는 Call-by-Reference

```python
x = [1, 2, 3]
y = x.copy()  # shallow copy
y[0] = 99
print(x, id(x))  # 2667623283392
print(y, id(y))  # 2667643787584

# deep copy 문제
x = [[1, 2], 3, 4]
y = x.copy()
y[1] = -1121
y[0][0] = 99
print(x, id(x))
print(y, id(y))
```

deepcopy까지 동원해야 할 정도로 중첩됐다면, list를 잘못 구성한 건 아닌지 의심해 볼 필요가 있다

```python
import copy

x = [[1, 2], 3, 4]
y = copy.deepcopy(x)  # 진짜 deep copy
y[0][0] = 99

print(x, id(x))
print(y, id(y))
```

```python
# 'append', 'extend'

x = []
x.append(3)
print(x, len(x))

x.append((3, 4))
print(x, len(x))

x.append("hello")
print(x, len(x))

x.append([9, 9, 9])
print(x, len(x))

###

x.extend([9, 9, 9])  # x += [9, 9, 9]
print(x, len(x))
```

```python
# 'insert', 'pop', 'remove', del

x = [3, 3, 3, 2, 1]
print(x)  # [3, 3, 3, 2, 1]
print(x.pop(-1))
print(x)  # [3, 3, 3, 2]

del x[0]  # x.remove(3)
print(x)

help(list.remove)

x.insert(1, 99)
print(x)
help(list.insert)
```

```python
# 'clear', 'count', 'index'

x = [1, 1, 1, 1, 1, 1, 1, 1, 2, 3, 4]
print(x.count(1))
print(x.index(4))

x.clear()
print(x)
```

# dictionary

> key-value 의 관계를 저장하는 자료형

```python
d = {}
print(d, type(d))

d = {'a': 1, 'b': 2, 'c': 3}
print(d, type(d))

print(d['a'])  # 키 값으로 접근
```

```python
d = {"a": 1, "b": 2, "c": 3}
print(d['b'])

d['c'] = 4  # update
print(d)

d['c'] = 3
d['d'] = 5  # append
print(d)
```

```python
print(dir({}))
```

```python
d = {"a": 1, "b": 2, "c": 3}
print('a' in d)
print(1 in d)  # 키 중심으로 찾는다

del d['b']
print(d)
```

```python
# 'keys', 'values'

d = {"a": 1, "b": 2, "c": 3}
print(d.keys())  # dict_keys(['a', 'b', 'c'])
print(d.values())  # dict_values([1, 2, 3])
print(d.items())  # dict_items([('a', 1), ('b', 2), ('c', 3)])

print(list(d.keys()))  # ['a', 'b', 'c']
print(list(d.values()))  # [1, 2, 3]
print(list(d.items()))  # [('a', 1), ('b', 2), ('c', 3)]

print(1 in d.values())  # 값을 찾기
```

```python
# 'get'

d = {'a': 1}
# print(d['b'])  # KeyError
print(d.get("a"))
print(d.get("b"))
print(d.get("c", "없을 때 출력할 메시지"))
```

```python
# 'pop', 'popitem'

d = {'a': 1, "b": 2, 'c': 3}
print(d)
print('---')
print(d.pop('a'))
print("---")
print(d.popitem())
print(d)
print("---")
print(d)
```

```python
help(dict.fromkeys)
```

```python
# 'clear', 'copy', 'update'
d = {'a': 1, 'b': 2, 'c': 3}
print(d)

d2 = {'c': 5, 'd': 4}
d = d | d2  # or 연산 (합집합)
print(d)

d.clear()
print(d)
```

```python
d = dict([
    ['a', 1],
    ['b', 2]
])
print(d)

d = dict([
    ("a", 1), ("b", 2)
])
print(d)

d = dict(
    (["a", 1], ["b", 2])
)
print(d)

d = dict(
    (("a", 1), ("b", 2))
)
print(d)

# 전부 다 {'a': 1, 'b': 2} 를 출력
```

```python
d = {"a": 1, "b": 1.3, "c": True,
     "d": None, "d": "Hello",
     }
print(d, type(d), len(d))

d = {"a": 1, "a": 2,
     "b": 3}
print(d)
```

```python
# key에는 불변객체만 들어갈 수 있다

d = {
    0: "a",
    1.4: "b",
    None: "c",
    (1, 3): "d",
    # [1, 4]: "e",  # unhashable type
}
print(d)
```

```python
print(hash(123))
print(hash(123.))

print(hash(123.0))

print(hash('hello'))
print(hash("world"))
print(hash(None))
print(hash(
    (1, 2)
))

print(type(123), type(123.))
d = {123: 'a', 123.: 'b'}
print(d)
```

```python
print(hash(0))  # 0
print(hash(0.0))  # 0
print(hash(False))  # 0
print(hash(""))  # 0
print(hash(None))  # 4238894112
print(hash(()))  # 5740354900026072187
# print(hash([]))
# print(hash({}))
# print(hash(set()))

d = {
    0: "a", 0.0: 'b', False: 'c',  # hash collision
    '': 'f',  # 키가 문자형일 때에도 해시 충돌이 발생하지만, 덮어씌워지진 않는다
    None: 'd',
    (): 'e',
}
print(d)
```

setdefault - 딕셔너리 shape

```python
d = {}
d.setdefault("one")  # d.setdefault("one", None)
print(d)

d.setdefault("four", 4)
print(d)
```

```python
# comprehension

l = [a for a in range(8) if a % 2 == 0]
print(l)

d = {a: a * 2 for a in "abcd"}
print(d)
```

```python
A = {1, 2, 3}
B = {2, 3, 4}
print(A ^ B)  # 대칭차집합 # symmetric_difference
print(A & B)
print(A | B)
print(A - B)  # 차집합

print('-' * 10)

A = {1, 2, 3}
B = {2, 3, 4}
A ^= B
A &= B
A |= B
A -= B
print(A)
```

```python
print({1, 2, 3}.isdisjoint({4, 5, 6}))
print({1, 2, 3}.issubset({1, 2, 3, 4}))  # 부분집합
print({1, 2, 3}.issuperset({1, 2}))
```
