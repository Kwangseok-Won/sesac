```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42884?language=python3

def solution(routes: list[list[int]]):
    routes.sort(key=lambda x: x[1])
    answer = 0
    # 초기 카메라 위치, 모든 경로의 시작점보다 작게 설정
    camera = -30001  # 아직 카메라 없음

    for route in routes:
        # 현재 경로의 시작점이 카메라 위치보다 크면 새로운 카메라 필요
        # 마지막 카메라보다 늦게 들어오면 못 찍음 -> 이 차 끝에 새 카메라
        if camera < route[0]:
            camera = route[1]
            answer += 1
    
    return answer


print(solution([[-20, -15], [-14, -5], [-18, -13], [-5, -3]]))  # 2
```
