```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12953?language=python3

def solution(arr: list[int]):
    answer = 0

    largest = max(arr)
    answer = largest

    while True:
        flag = True
        for num in arr:
            if answer % num != 0:
                flag = False
                break
        if flag:
            return answer
        answer += largest

    return answer


print(solution([2, 6, 8, 14]))  # 168
print(solution([1, 2, 3]))  # 6
```

```python
def solution2(arr: list[int]) -> int:
    largest = max(arr)
    answer = largest


    while True:
        if all(answer % num == 0 for num in arr):
            return answer
        answer += largest


print(solution2([2, 6, 8, 14]))  # 168
print(solution2([1, 2, 3]))  # 6
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42885?language=python3

def solution(people: list[int], limit: int) -> int:
    answer = 0

    people.sort()

    min_idx = 0
    max_idx = len(people) - 1
    while min_idx <= max_idx:
        if people[min_idx] + people[max_idx] <= limit:
            min_idx += 1
        max_idx -= 1

        answer += 1

    return answer


print(solution([70, 50, 80, 50], 100))  # 3
print(solution([70, 80, 50], 100))  # 3
```

```python
# perceptron

X = [
    [0.8, 0.6],
    [0.6, 0.9],
    [0.1, 0.2],
    [0.3, 0.1],
    [0.6, 0.6],
    [0.4, 0.3],
    [0.1, 0.2],
]
Y = [1, 1, 0, 0, 1, 0, 0]
w_1, w_2, lr = 0.3, 0.2, 0.01


def step(s):
    """
    초창기(퍼셉트론)에 도입했던 계단 함수
    - 미분 불가
    - XOR 문제 해결 불가
    """
    return 1 if s >= 0.5 else 0


epoch = 30
for _ in range(epoch):
    wrong = 0

    for (x_1, x_2), y in zip(X, Y):
        s = (x_1 * w_1) + (x_2 * w_2)
        y_predict = step(s)
        error = y - y_predict

        w_1 = w_1 + x_1 * error * lr
        w_2 = w_2 + x_2 * error * lr
        wrong += abs(error)

        print(f"epoch: {_ + 1}, w_1: {w_1:.4f}, w_2: {w_2:.4f}, wrong: {wrong}")
    if wrong == 0:
        break
```

```python
import math

# 1. 데이터셋
# x_1, x_2: 입력값, Y: 정답 (1=비, 0=맑음)
X = [
    (0.8, 0.6),
    (0.6, 0.9),
    (0.1, 0.2),
    (0.3, 0.1),
    (0.6, 0.6),
    (0.4, 0.3),
    (0.1, 0.2),
]
Y = [1, 1, 0, 0, 1, 0, 0]

# 2. 초기값
w_1, w_2 = 0.3, 0.2  # 초기 연결강도
bias = -0.5
lr = 0.01  # 학습률
n = len(X)

MAX_EPOCH = 30  # 최대 반복 횟수 (데이터 전체를 한 바퀴 도는 것 = 1 epoch)


def sigmoid(z):
    return 1 / (1 + math.exp(-z))


# 3. 학습 과정을 한 단계씩 내보내는 함수
def train_steps():
    global w_1, w_2, bias

    for epoch in range(1, MAX_EPOCH + 1):
        y_predict = [sigmoid((w_1 * x_1) + (w_2 * x_2) + bias) for x_1, x_2 in X]
        error = [y_p - y_i for y_p, y_i in zip(y_predict, Y)]

        # ∂E/∂w_1, ∂E/∂w_2, ∂E/∂bias
        dw_1 = 2 * sum(err * x_1 for err, (x_1, x_2) in zip(error, X)) / n
        dw_2 = 2 * sum(err * x_2 for err, (x_1, x_2) in zip(error, X)) / n
        d_bias = 2 * sum(error) / n

        w_1 -= lr * dw_1
        w_2 -= lr * dw_2
        bias -= lr * d_bias

        print(
            f"epoch {epoch}: w_1: {w_1:.6f}, w_2: {w_2:.6f}, bias: {bias:.6f}, error: {sum(error):.6f}"
        )


train_steps()
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181879?language=python3

def solution(num_list: list[int]) -> int:
    answer = 0

    length = len(num_list)
    if length >= 11:
        return sum(num_list)

    mul = 1
    if length <= 10:
        for num in num_list:
            mul *= num
    return mul

    return answer


print(solution([3, 4, 5, 2, 5, 4, 6, 7, 3, 7, 2, 2, 1]))  # 51
print(solution([2, 3, 4, 5]))  # 120
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181878?language=python3

def solution(myString: str, pat: str) -> int:
    answer = 0

    if len(myString) < len(pat):
        return 0

    myString = myString.lower()
    pat = pat.lower()

    gap = len(pat)
    for idx in range(len(myString)):
        end = idx + gap
        if end <= len(myString) and myString[idx:end] == pat:
            return 1

    return answer


print(solution("AbCdEfG", "aBc"))  # 1
print(solution("aaAA", "aaaaa"))  # 0
```
