```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12953?language=python3

def solution(arr: list[int]):
    answer = 0

    largest = max(arr)
    answer = largest

    while True:
        flag = True
        for num in arr:
            if answer % num != 0:
                flag = False
                break
        if flag:
            return answer
        answer += largest

    return answer


print(solution([2, 6, 8, 14]))  # 168
print(solution([1, 2, 3]))  # 6
```

```python
def solution2(arr: list[int]) -> int:
    largest = max(arr)
    answer = largest


    while True:
        if all(answer % num == 0 for num in arr):
            return answer
        answer += largest


print(solution2([2, 6, 8, 14]))  # 168
print(solution2([1, 2, 3]))  # 6
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/42885?language=python3

def solution(people: list[int], limit: int) -> int:
    answer = 0

    people.sort()

    min_idx = 0
    max_idx = len(people) - 1
    while min_idx <= max_idx:
        if people[min_idx] + people[max_idx] <= limit:
            min_idx += 1
        max_idx -= 1

        answer += 1

    return answer


print(solution([70, 50, 80, 50], 100))  # 3
print(solution([70, 80, 50], 100))  # 3
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181879?language=python3

def solution(num_list: list[int]) -> int:
    answer = 0

    length = len(num_list)
    if length >= 11:
        return sum(num_list)

    mul = 1
    if length <= 10:
        for num in num_list:
            mul *= num
    return mul

    return answer


print(solution([3, 4, 5, 2, 5, 4, 6, 7, 3, 7, 2, 2, 1]))  # 51
print(solution([2, 3, 4, 5]))  # 120
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181878?language=python3

def solution(myString: str, pat: str) -> int:
    answer = 0

    if len(myString) < len(pat):
        return 0

    myString = myString.lower()
    pat = pat.lower()

    gap = len(pat)
    for idx in range(len(myString)):
        end = idx + gap
        if end <= len(myString) and myString[idx:end] == pat:
            return 1

    return answer


print(solution("AbCdEfG", "aBc"))  # 1
print(solution("aaAA", "aaaaa"))  # 0
```
