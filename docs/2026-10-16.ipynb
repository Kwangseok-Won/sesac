```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181891?language=python3

def solution(num_list: list[int], n: int) -> list[int]:
    answer = []

    head = num_list[n:]
    tail = num_list[:n]

    answer.extend(head)
    answer.extend(tail)
    
    return answer


print(solution([2, 1, 6], 1))  # [1, 6, 2]
print(solution([5, 2, 1, 7, 5], 3))  # [7, 5, 5, 2, 1]
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181887?language=python3

def solution(num_list: list[int]) -> int:
    answer = 0

    even = sum(num_list[::2])
    odd = 0

    for idx in range(len(num_list)):
        if idx % 2 != 0:
            odd += num_list[idx]

    answer = max(even, odd)

    return answer


print(solution([4, 2, 6, 1, 7, 6]))  # 17
print(solution([-1, 2, 5, 6, 3]))  # 8
```
