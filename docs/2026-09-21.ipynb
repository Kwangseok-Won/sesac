튜플을 쓸 때 — 바뀌면 안 되는 값. 좌표 (위도, 경도), RGB 색상 (255, 0, 0), 요일 이름, 설정값처럼 고정된 묶음.
리스트를 쓸 때 — 추가·삭제·수정이 필요한 값. 할 일 목록, 장바구니, 점수 기록처럼 변하는 데이터.

튜플의 장점은 4a에서 말한 불변의 장점과 같습니다 — 안전하고(실수로 안 바뀜), 빠르며, 다음다음 챕터에서 배울 딕셔너리의 열쇠로도 쓸 수 있습니다 (리스트는 열쇠로 못 씁니다). 로봇 제어에서 "고정된 관절 순서"나 "센서 좌표"처럼 바뀌면 안 되는 묶음에 튜플이 잘 맞아요.

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time() # pre-hook
        result = func(*args, **kwargs)
        end = time.time() # post-hook

        print(f"실행 시간: {end - start:.4f}초")
        return result
    return wrapper
```

```python
# https://ko.wikipedia.org/wiki/에라토스테네스의_체

@timer
def prime_definition(n):
    for i in range(2, n + 1):
        detector = 0
        for k in range(2, i):
            if i % k == 0:
                detector += 1
        if detector == 0:
            print(i)

prime_definition(400)
```

```python
@timer
def prime_printer(n):
    prime_list = []
    for i in range(2, n + 1):
        detector = 0
        for k in prime_list:
            if i % k == 0:
                detector += 1
        if detector == 0:
            prime_list.append(i)
            print(i)

prime_printer(400)
```

```python
def find_mode():
    nums = list(map(int, input("정수들을 공백으로 입력하세요: ").split()))
    freq = {}

    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1

    max_freq = max(freq.values())

    for key in freq:
        if freq[key] == max_freq:
            print("최빈값: ", key)
            break # 여러 개이면 그 중 하나만 출력


find_mode()
```

```python
# 정렬하지 않고, 두 번째로 큰 수 찾기
def find_second_largest():
    nums = list(map(int, input("정수들을 공백으로 입력하세요: ").split()))

    first = second = float("-inf")

    for num in nums:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    if second == float("-inf"):
        print("두 번째로 큰 수가 없습니다.")
    else:
        print("두 번째로 큰 수: ", second)


find_second_largest()
```

```python
# https://en.wikipedia.org/wiki/Run-length_encoding#Python_implementation
def run_length_encoding(text):
    if not text:
        return ""

    result = ""
    prev_char = text[0]
    count = 1

    for i in range(1, len(text)):
        if text[i] == prev_char:
            count += 1
        else:
            result += prev_char + (str(count) if count > 1  else "")
            prev_char = text[i]
            count = 1

    result += prev_char + (str(count) if count > 1 else "")
    return result


def runner():
    s = input("영문자열을 입력하세요: ")
    encoded = run_length_encoding(s)
    print("압축 결과: ", encoded)


runner()
```

```python

def is_valid_brackets(s: str):
    stack = []
    pair = {
        ")": "(",
        "}": "{",
        "]": "["
    }

    for char in s:
        if char in "({[":
            stack.append(char)
        elif char in ")}]":
            if not stack or stack [-1] != pair[char]:
                return False
            stack.pop()
    return not stack


def runner():
    expr = input("괄호 문자열을 입력하세요: ")
    print("올바른 괄호 구조: ", is_valid_brackets(expr))


runner()
```

```python
# 일정 관리 콘솔 프로그램 (CRUD)

def print_menu():
    """
    메뉴 출력
    :return:
    """
    print("\n 일정 관리 시스템")
    print("1. 일정 등록")
    print("2. 일정 조회")
    print("3. 일정 수정")
    print("4. 일정 삭제")
    print("5. 종료")

def schedule_manager():
    schedules: dict[str, str] = {}

    while True:
        print_menu()
        choice = input("메뉴를 선택하세요 (1-5): ").strip()

        if choice == "1":
            date = input("날짜 입력 (예: 2026-09-21): ").strip()
            content = input("일정 내용 입력: ").strip()
            schedules[date] = content
            print("일정이 등록되었습니다.")

        elif choice == "2":
            date = input("조회할 날짜 입력: ").strip()
            if date in schedules:
                print(f"{date}: {schedules[date]}")
            else:
                print("해당 날짜의 일정이 없습니다.")

        elif choice == "3":
            date = input("수정할 날짜 입력: ").strip()
            if date in schedules:
                new_content = input("새 일정 내용 입력: ").strip()
                schedules[date] = new_content
                print("일정이 수정되었습니다.")
            else:
                print("수정할 일정이 존재하지 않습니다.")

        elif choice == "4":
            date = input("삭제할 날짜 입력: ").strip()
            if date in schedules:
                del schedules[date]
                print("일정이 삭제되었습니다.")
            else:
                print("삭제할 일정이 존재하지 않습니다.")

        elif choice == "5":
            print("일정 관리 프로그램을 종료합니다.")
            break

        else:
            print("유효한 메뉴 번호를 입력하세요.")


schedule_manager()
```

```python

def die(a: int, b: int, c: int) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181930?language=python3
    :param a:
    :param b:
    :param c:
    :return:
    """

    if a == b == c:
        return (a+b+c)*(a**2 + b**2 + c**2)*(a**3 + b**3 + c**3)
    elif a == b != c or a == c != b or b == c != a:
        return (a+b+c) * (a**2 + b**2 + c**2)
    else:
        return a + b + c

    return

print(die(2, 6, 1))
print(die(5, 3, 3))
print(die(4, 4, 4))
```

```python
def solution(num_list: list[int]) -> int:
    """
    https://school.programmers.co.kr/learn/courses/30/lessons/181929?language=python3
    :param num_list:
    :return:
    """
    answer = 0

    multi_each = 1
    sum_each = 0

    for num in num_list:
        multi_each *= num
        sum_each += num

    if multi_each < sum_each ** 2:
        return 1
    return 0

print(solution([3, 4, 5, 2, 1]))
print(solution([5, 7, 8, 3]))
```

```python
def solution(num_list: list[int]) -> list[int]:
    answer = []

    last = num_list[::-len(num_list)].pop()
    pre_last = num_list[len(num_list) - 2]

    if last > pre_last:
        num_list.append(last - pre_last)
        return num_list
    num_list.append(last * 2)
    return num_list

    return answer

print(solution([2, 1, 6]))
print(solution([5, 2, 1, 7, 5]))
```
