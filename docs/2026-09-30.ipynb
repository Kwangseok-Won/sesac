```python
import pandas as pd

# Series: 1차원 데이터
# 가로로 뻗어나가는 축 하나: axis 0

s1 = pd.Series([5000, 30000, 20000, 15000])
display(s1)

print("-" * 40)

s2 = pd.Series({\
    "2020-03-03": 5000, "2020-03-04": 30000,\
    "2020-03-05": 20000, "2020-03-06": 15000
})
display(s2)

print("-" * 40)

# 데이터와 인덱스를 통으로 입력
s3 = pd.Series([5000, 30000, 20000, 15000],\
               index=["2020-03-03", "2020-03-04", "2020-03-05", "2020-03-06"],
               name="durian")
s3.index.name = "dates"

# 상동
s3 = pd.Series([5000, 30000, 20000, 15000])
# attribute로 접근
s3.index = ("2020-03-03", "2020-03-04", "2020-03-05", "2020-03-06")
s3.name = "durian"
display(s3)

print("-" * 40)

s = pd.Series([1, 2, 3, 4, 5])
display(s)
display(s.values)
display(s.index)
```

```python

df2 = pd.DataFrame([
    [4000, 20000, 10000],
    [5000, 8000, 11000],
    [6000, 9000, 12000]
])
df2.index = [[
    "apple",
    "banana",
    "cherry"
]]
df2.columns = [["03.03", "03.04", "03.05"]]
display(df2)

print("-" * 40)
print("시리즈를 합친 것이 데이터프레임")
s1 = pd.Series(
    [4000, 20000, 10000],
    index=["03.03", "03.04", "03.05"],
    name="apple"
)
s2 = pd.Series(
    [5000, 8000, 11000],
    index=["03.03", "03.04", "03.05"],
    name="banana"
)
s3 = pd.Series(
    [6000, 9000, 12000],
    index=["03.03", "03.04", "03.05"],
    name="cherry"
)
dataframe_from_series = pd.DataFrame([s1, s2, s3])
display(dataframe_from_series)
```

```python
# https://blog.naver.com/youji4ever/221712578078

dates = pd.date_range("20130101", periods=6)
df = pd.DataFrame(np.random.randn(6,4), index=dates, columns=list("ABCD"))
display(df)

print("-" * 40)

# reindex(재 인덱스)는 해당 축에 대하여 인덱스를 변경/추가/삭제
df1 = df.reindex(index=dates[0:4], columns=list(df.columns) + ["E"])
df1.loc[dates[0]:dates[1], "E"] = 1
display(df1)

print("-" * 40)

display(pd.isna(df1))
display(df1.dropna(how="any"))
display(df1.fillna(value=5))
```

```python
df1 = pd.DataFrame({
    "Favorite Food": [
        "Alfredo Pasta",
        "Bobba",
        "Croissants",
        "Danish Rolls",
        "Edaame",
        "Fish",
        "Ice Cream"
    ]
    },
        index=pd.Index(["Alfred", "Bob", "Cris", "Daniel", "Edward", "Fitz", "Ivan"],
        name="Name"),
)
df2 = pd.DataFrame({
    "Favorite Country": [
        "Albania",
        "Croatia",
        "Denmark",
        "Fiji",
        "Germany",
        "Haiti"
    ]},
    index=pd.Index(["Alfred", "Cris", "Daniel", "Fitz", "Gerald", "Hayden"],
    name="Name"),
)
display(df1)
display(df2)
print("-" * 40)

# axis가 없으면: 세로로 길쭉하게 합치기
display(pd.concat([df1, df2], sort=False))  # concat 사용 <- append는 pandas 2.0부터 deprecated
```

```python
display(pd.concat([df1, df2], axis=1, sort=False))  # axis=1 -> 열 추가(옆으로 붙이기)
```

```python
display(pd.concat([df1, df2], axis=1, sort=True))  # Fitz 다음 Gerald, Hayden, Ivan 순으로 정렬됐다
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/181924?language=python3

def solution(arr: list[int], queries: list[list[int, int]]) -> list[int]:
    answer = []

    for i in range(len(queries)):
        first_query, second_query = queries[i]

        arr[first_query], arr[second_query] = arr[second_query], arr[first_query]

    return answer
```
