```python
# https://school.programmers.co.kr/learn/courses/30/lessons/43165?language=python3

def solution(numbers: list[int], target: int) -> int:
    answer = 0

    def dfs(index: int, total: int) -> None:
        print(f"dfs called with index={index}, total={total}")
        nonlocal answer
        print(f"current answer={answer}")
        if index == len(numbers):
            print(f"reached end with total={total}")
            if total == target:
                print(f"total matches target, incrementing answer")
                answer += 1
                print(f"new answer={answer}")
            return
        print(f"going deeper with index={index}, total={total}")
        dfs(index + 1, total + numbers[index])
        print(f"going deeper with index={index}, total={total}")
        dfs(index + 1, total - numbers[index])

    print(f"starting dfs with numbers={numbers}, target={target}")
    dfs(0, 0)
    print(f"finished dfs with final answer={answer}")

    return answer


print(solution([1, 1, 1, 1, 1], 3))  # 5
print(solution([4, 1, 2, 1], 4))  # 2
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12912?language=python3

def solution(a: int, b: int) -> int:
    answer = 0

    start = min(a, b)
    stop = max(a, b) + 1

    for value in range(start, stop):
        answer += value
    
    return answer


print(solution(3, 5))  # 12
print(solution(3, 3))  # 3
print(solution(5, 3))  # 12
```

```python
# https://school.programmers.co.kr/learn/courses/30/lessons/12903?language=python3

def solution(s: str) -> str:
    answer = ""
    n = len(s)

    if n % 2 == 0:
        return s[n // 2 - 1: n // 2 + 1]
    else:
        return s[n // 2]
    
    return answer


print(solution("abcde"))  # "c"
print(solution("qwer"))  # "we"
```

```python
from collections import deque


def solution(friends, start, end):
    graph = {}
    for a, b in friends:
        graph.setdefault(a, []).append(b)
        graph.setdefault(b, []).append(a)

    dist = {start: 0}
    dq = deque([start])
    while dq:
        now = dq.popleft()
        if now == end:
            return dist[now]
        for nxt in graph[now]:
            for nxt not in dist:
                dist[nxt] = dist[now] + 1
                dq.append(nxt)

    return -1
```
