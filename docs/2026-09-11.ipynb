```python
matrix = [[1, 2], [3, 4], [5, 6]]

rows = len(matrix)
cols = len(matrix[0])
print(f"rows: {rows}, cols: {cols}")

transposed = []

# 여기에 코드를 작성하세요
for i in zip(*matrix):
    transposed.append(list(i))

print(f"transposed: {transposed}")

# https://docs.python.org/3/reference/expressions.html#starred-subscriptions
# https://peps.python.org/pep-3132/
print("*matrix: ", *matrix)
```
