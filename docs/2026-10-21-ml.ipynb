```python
import math

# 1. 데이터셋
# x_1, x_2: 입력값, Y: 정답 (1=비, 0=맑음)
X = [
    (0.8, 0.6),
    (0.6, 0.9),
    (0.1, 0.2),
    (0.3, 0.1),
    (0.6, 0.6),
    (0.4, 0.3),
    (0.1, 0.2),
]
Y = [1, 1, 0, 0, 1, 0, 0]

# 2. 초기값
w_1, w_2, w_3, w_4, w_5, w_6 = 0.7, 0.3, 0.4, 0.6, 0.55, 0.45  # 초기 연결강도
x_1, x_2 = 0.8, 0.6
bias = -0.5
lr = 0.01  # 학습률


def sigmoid(z):
    return 1 / (1 + math.exp(-z))


epoch = 3000
for _ in range(epoch):
    for (x_1, x_2), y in zip(X, Y):
        # 순전파
        h_1 = sigmoid(x_1*w_1 + x_2*w_3)
        h_2 = sigmoid(x_1*w_2 + x_2*w_4)
        o = sigmoid(h_1*w_5 + h_2*w_6)

        # Backpropagation
        d_3 = o - y  # 시그모이드 + BCE 출력층의 미분계수
        d_1 = d_3 * w_5 * h_1*(1-h_1)
        d_2 = d_3 * w_6 * h_2*(1-h_2)

        # 업데이트
        w_5 -= lr*d_3*h_1; w_6 -= lr*d_3*h_2
        w_1 -= lr*d_1*x_1; w_3 -= lr*d_1*x_2
        w_2 -= lr*d_2*x_1; w_4 -= lr*d_2*x_2

        print(f"h_1: {h_1}, h_2: {h_2}, o: {o}, d_1: {d_1}, d_2: {d_2}, d_3: {d_3}")
```
